### リスト

まず、`scores`という変数に、5つの点数をリストとして代入しています。リストを使うことで、複数のデータをひとまとめにして管理することができます。

次に、`print(scores)`によってリスト全体を表示しています。その後、`scores[0]`によってリストの最初の要素を取り出して表示しています。Pythonでは、リストの要素を0から数えるため、`scores[0]`は「70」を表します。

```python
print(sum(scores))
print(len(scores))
```

## 4. 条件分岐・繰り返しを体験する

### 条件分岐

まず、変数`score`に「80」を代入しています。次に、`if`文を使って、点数が60点以上かどうかを判定しています。

`score >= 60`という条件が成立する場合は、「合格」と表示します。一方、条件が成立しない場合は、`else`の処理が実行され、「不合格」と表示されます。この例では、点数が80点なので条件が成立し、「合格」と表示されます。

このように、`if`文を使うことで、データの値や条件に応じて処理を切り替えることができます。条件分岐は、データ分析や機械学習だけでなく、さまざまなプログラムで利用される基本的な仕組みです。

```python
score = 80

if score >= 60:
    print("合格")
else:
    print("不合格")
```

```python
score = 50

if score >= 60:
    print("合格")
else:
    print("不合格")
```

### 繰り返し

まず、`scores`という変数に5つの点数をリストとして保存しています。次に、`for`文を使って、リストの中にある点数を1つずつ取り出しています。

`for score in scores:`では、リスト`scores`から要素を順番に取り出し、その値を変数`score`に代入します。その後、`print(score)`によって、取り出した点数を画面に表示しています。この処理がリストのすべての要素に対して繰り返されるため、「70」「85」「90」「65」「80」が順番に表示されます。

このように、`for`文を使うと、リストやその他の複数のデータを1つずつ取り出して、同じ処理を繰り返すことができます。データ分析では、複数のデータを順番に処理するときに利用される基本的な仕組みです。

```python
scores = [70, 85, 90, 65, 80]

for score in scores:
    print(score)
```

## 5. Pythonで簡単なデータを扱ってみる

まず、`scores`というリストに5人分の点数を保存しています。`len`、`sum`、`max`、`min`などの関数を使うことで、このリストからさまざまな集計値を求めることができます。

`len(scores)`はリストの要素数を求めるため、「5」となります。`sum(scores)`は点数の合計を求めるため、「390」となります。また、`sum(scores) / len(scores)`によって合計点を人数で割り、平均点を計算しています。さらに、`max(scores)`で最高点を、`min(scores)`で最低点を求めています。

この例では、人数が5人、合計が390点、平均が78.0点、最高点が90点、最低点が65点となります。このように、Pythonの基本的な関数を組み合わせることで、リストに保存されたデータからさまざまな統計量を簡単に計算できます。

```python
scores = [70, 85, 90, 65, 80]

print("人数:", len(scores))
print("合計:", sum(scores))
print("平均:", sum(scores) / len(scores))
print("最高点:", max(scores))
print("最低点:", min(scores))
```

まず、`scores`に5人分の点数を保存し、`sum(scores) / len(scores)`によって平均点を計算しています。計算した平均点は、変数`average`に保存しています。この例では、平均点は78.0点になります。

次に、`for`文を使ってリストから点数を1つずつ取り出し、`if`文で平均点と比較しています。`score >= average`という条件が成立すれば「平均以上」、成立しなければ「平均未満」と表示します。

この例では、70点と65点は平均未満、85点、90点、80点は平均以上と判定されます。このように、集計によって求めた値を基準として条件分岐を行うことで、データを分類したり、特徴を調べたりすることができます。


```python
scores = [70, 85, 90, 65, 80]

average = sum(scores) / len(scores)

for score in scores:
    if score >= average:
        print(score, "平均以上")
    else:
        print(score, "平均未満")
```
