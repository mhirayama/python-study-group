```python
score = 80

if score >= 60:
    print("合格")
else:
    print("不合格")
```

## 5. Pythonで簡単なデータを扱ってみる

```python
scores = [70, 85, 90, 65, 80]

print("人数:", len(scores))
print("合計:", sum(scores))
print("平均:", sum(scores) / len(scores))
print("最高点:", max(scores))
print("最低点:", min(scores))
```

```python
scores = [70, 85, 90, 65, 80]

average = sum(scores) / len(scores)

for score in scores:
    if score >= average:
        print(score, "平均以上")
    else:
        print(score, "平均未満")
```
