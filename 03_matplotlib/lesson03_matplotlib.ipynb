# 1. pandasのデータをグラフにする

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV data
df = pd.read_csv("sales.csv")

# Calculate sales
df["sales"] = df["quantity"] * df["price"]

print(df)

# Plot the data
df.plot(x="product", y="sales", kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

# 2. 棒グラフで比較する

```python
# Aggregate sales by product
sales_by_product = df.groupby("product")["sales"].sum()

print(sales_by_product)

# Create a bar chart
sales_by_product.plot(kind="bar")

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

# 3. 折れ線グラフで変化を見る

```python
# Convert date to datetime
df["date"] = pd.to_datetime(df["date"])

# Aggregate sales by date
sales_by_date = df.groupby("date")["sales"].sum()

print(sales_by_date)

# Create a line chart
sales_by_date.plot(kind="line", marker="o")

plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

# 4. グラフからデータの特徴を読み取る

```python
# Find the product with the highest sales
top_product = sales_by_product.idxmax()
top_sales = sales_by_product.max()

print("Top product:", top_product)
print("Top sales:", top_sales)

# Find the date with the highest sales
top_date = sales_by_date.idxmax()
top_daily_sales = sales_by_date.max()

print("Top sales date:", top_date)
print("Top daily sales:", top_daily_sales)
```
