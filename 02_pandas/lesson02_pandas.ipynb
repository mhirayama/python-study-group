# 1．データを読み込む

以下のコードは、Pythonのライブラリであるpandasを使って、CSVファイルに保存されたデータを読み込み、画面に表示するものです。

最初にpandasを読み込み、CSVファイルをデータフレームとして取得しています。データフレームは、表形式のデータを扱うためのpandasの基本的なデータ構造です。

ここでは「sales.csv」というCSVファイルを読み込み、その内容を画面に表示しています。


```python
import pandas as pd

df = pd.read_csv("sales.csv")

print(df)
```

どのような項目がデータに含まれているのかを確認できるため、データを分析するときの最初の確認として便利です。

# 3．データを整理する

以下のコードは、既存のデータから計算を行い、その結果を新しい列として追加するものです。

ここでは、商品の数量と価格を掛け合わせて売上金額を計算し、「sales」という列に追加しています。最後にデータフレーム全体を表示して、追加された列を確認しています。


```python
df["sales"] = df["quantity"] * df["price"]

print(df)
```

以下のコードは、データフレームから特定の列を取り出して表示するものです。

ここでは「product」列を選択し、その列に含まれるデータを表示しています。特定の項目だけを確認したい場合に便利です。


```python
print(df["product"])
```

以下のコードは、データフレームから特定の条件に一致するデータだけを抽出するものです。

ここでは、「product」列の値が「Apple」の行だけを取り出し、「apple」という変数に格納しています。条件に合うデータを絞り込んで確認したいときに便利です。


```python
apple = df[df["product"] == "Apple"]

print(apple)
```

# 4．データを集計する

以下のコードは、商品ごとに数量をまとめて集計するものです。

ここでは、商品名ごとに数量をグループ化し、それぞれの合計数量を計算しています。商品ごとの販売数などを集計するときに便利です。


```python
quantity_by_product = df.groupby("product")["quantity"].sum()

print(quantity_by_product)
```

以下のコードは、商品ごとの売上金額を集計するものです。

ここでは、商品名ごとにデータをグループ化し、「sales」列の合計を計算しています。商品ごとの売上を比較するときなどに便利です。


```python
sales_by_product = df.groupby("product")["sales"].sum()

print(sales_by_product)
```

以下のコードは、商品ごとの平均価格を計算するものです。

ここでは、商品名ごとにデータをグループ化し、「price」列の平均値を計算しています。商品ごとの平均価格を比較するときなどに便利です。


```python
average_price = df.groupby("product")["price"].mean()

print(average_price)
```

以下のコードは、商品ごとに複数の項目をまとめて集計するものです。

ここでは、商品の数量と売上金額の合計、価格の平均を一度に計算し、「summary」というデータフレームにまとめています。複数の指標をまとめて確認したいときに便利です。


```python
summary = df.groupby("product").agg(
    total_quantity=("quantity", "sum"),
    total_sales=("sales", "sum"),
    average_price=("price", "mean")
)

print(summary)
```

以下のコードは、CSVファイルを読み込み、データの内容を確認したうえで、商品ごとの販売数量、売上金額、平均価格を集計するものです。

まずpandasを使ってCSVデータを読み込み、データの先頭部分や行数・列数、データの構造を確認しています。

次に、数量と価格から売上金額を計算し、新しい列として追加しています。その後、商品ごとにデータをまとめ、販売数量と売上金額の合計、価格の平均を計算しています。

このように、pandasを使うと、CSVファイルの読み込みからデータの確認、計算、集計までをまとめて行うことができます。


```python
import pandas as pd

# Load CSV data
df = pd.read_csv("sales.csv")

# Check the data
print(df.head())
print(df.shape)
print(df.info())

# Add a sales column
df["sales"] = df["quantity"] * df["price"]

# Aggregate the data
summary = df.groupby("product").agg(
    total_quantity=("quantity", "sum"),
    total_sales=("sales", "sum"),
    average_price=("price", "mean")
)

print(summary)
```
